Visualization 1: This visualization helps describe the relationship between seed and record in the regular season. It indicates that teams seeded higher (1 is the highest ranked) often have a better team record. It looks like there is a weak positive relationship between these two variables, representing the fact that other factors other than regular season wins are used to seed teams in the four brackets. The four years used selected start at the most recent year in the data, 2018, and go back 10 years until 1988. 

```r
ncaa_data_2018 <- filter(ncaa_data, year == "2018")
vizualization_1a <- ggplot(ncaa_data_2018, aes(x = seed, y = reg_wins)) + 
geom_point() + 
labs(x = "Seed", y = "Regular Season Wins", title = "Regular Season Wins vs Tournament Seed (2018)") 
vizualization_1a

ncaa_data_2008 <- filter(ncaa_data, year == "2008")
vizualization_1b <- ggplot(ncaa_data_2008, aes(x = seed, y = reg_wins)) + 
geom_point() + 
labs(x = "Seed", y = "Regular Season Wins", title = "Regular Season Wins vs Tournament Seed(2008)") 
vizualization_1b

ncaa_data_1998 <- filter(ncaa_data, year == "1998")
vizualization_1c <- ggplot(ncaa_data_1998, aes(x = seed, y = reg_wins)) + 
geom_point() + 
labs(x = "Seed", y = "Regular Season Wins", title = "Regular Season Wins vs Tournament Seed(1998") 
vizualization_1c

ncaa_data_1988 <- filter(ncaa_data, year == "1988")
vizualization_1d <- ggplot(ncaa_data_1988, aes(x = seed, y = reg_wins)) + 
geom_point() + 
labs(x = "Seed", y = "Regular Season Wins", title = "Regular Season Wins vs Tournament Seed(1988") 
vizualization_1d

```

Visualization 2: Comparing seed and tournament finish is relevant to this analysis because seed is often one of the most popular predictors for the March Madness tournament. Seed is how the teams in the tournament are ranked in order to figure out how teams progress through the rounds (who they will play). There is a trend in all of the graphs below that indicate that higher seeded teams (1 is the highest) will stay in the tournament longer. In all four years, the winner was the first or second seed.

```r
ggplot(ncaa_data_1988, aes(x = tourney_finish, y = seed, fill = seed)) +
geom_bar(stat = "Identity") + 
       labs(x = "Round Lost in Tournament", y = "Seed", title = "Seed vs Round Out in March Madness (1988)") 

ggplot(ncaa_data_1998, aes(x = tourney_finish, y = seed, fill = seed)) +
geom_bar(stat = "Identity") + 
       labs(x = "Round Lost in Tournament", y = "Seed", title = "Seed vs Round Out in March Madness (1998)") 

ggplot(ncaa_data_1988, aes(x = tourney_finish, y = seed, fill = seed)) +
geom_bar(stat = "Identity") + 
       labs(x = "Round Lost in Tournament", y = "Seed", title = "Seed vs Round Out in March Madness(2008)") 

ggplot(ncaa_data_1988, aes(x = tourney_finish, y = seed, fill = seed)) +
geom_bar(stat = "Identity") + 
       labs(x = "Round Lost in Tournament", y = "Seed", title = "Seed vs Round Out in March Madness(2018)") 
       
```
